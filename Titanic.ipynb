# 🚢 Titanic Dataset Analysis Project
Welcome! This notebook explores the famous Titanic dataset to uncover patterns in passenger survival rates.

## Libraries

```python
from pickletools import uint8

import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from matplotlib.pyplot import xlabel, ylabel
from sklearn.preprocessing import StandardScaler,MinMaxScaler,LabelEncoder,OneHotEncoder
from torch.fx.experimental.unification.unification_tools import groupby
```

## 📥 Data Extraction
Loading the Titanic dataset using Seaborn and selecting the target columns for our analysis (`sex`, `age`, `pclass`, `survived`, `embarked`, `who`).

```python
original_df = sns.load_dataset("titanic")
original_df.info()
df = original_df[["sex","age","pclass","survived","embarked","who"]].copy()
#New Feature
df["embarked"] = df["embarked"].replace({
    'S': 'Southampton (UK)',
    'C': 'Cherbourg (France)',
    'Q': 'Queenstown (Ireland)'})

survived_male =df[df["who"]=="man"].groupby("embarked")["survived"].mean()

survivors_percentage = df.groupby("who")["survived"].mean()*100

financial_situation_of_the_survivors = df.groupby("pclass")["survived"].mean()





df.info()
df.describe()
df.head(20)
#print(df["who"].value_counts())
```

## 🧹 Data Cleaning & Preparation
Handling missing values, cleaning up data types, and preparing the dataset for accurate analysis and visualization.

```python
#Inplace NaN
df["age"] = df["age"].fillna(df.groupby("who")["age"].transform("mean"))
df["embarked"] = df["embarked"].fillna(df["embarked"].mode()[0])
df.info()

#Data Type Optimization
df["age"] = df["age"].astype("uint8")
df["pclass"] = df["pclass"].astype("uint8")
df["survived"] = df["survived"].astype("uint8")
df.info()
```

## 📊 Data Visualization & Insights
Visualizing our findings—including the impact of socio-economic status (`pclass`) on survival rates using Matplotlib and Seaborn.

```python
fig = plt.figure()

fig.add_subplot(2,2,1)
survived_male.plot(kind="bar", )
plt.xlabel("Embarked")
plt.ylabel("Survived_male")
plt.title("The survivors of any nationality")


fig.add_subplot(2,2,2)
colors=["lightgreen","salmon","lightblue"]
survivors_percentage.plot(kind="bar", color=colors)
for i,v in enumerate(survivors_percentage):
    plt.text(i, v , f"{v:.1f}%", ha="center",weight="bold")

plt.title("Percentage of survivors")

fig.add_subplot(2,2,3)
plt.bar(financial_situation_of_the_survivors.index,financial_situation_of_the_survivors.values)
plt.xlabel("pclass")
plt.ylabel("survived")
plt.title("The impact of financial situation on survival")
plt.show()
# ==========================================
# DATA EXPLANATION: Port of Embarkation
# ==========================================
# The 'embarked' column indicates the port where each passenger boarded the Titanic.
# There are three historical ports of embarkation:
#
# 1. Southampton (UK) - Represented as 'S':
#    - The main starting point of the Titanic's voyage.
#    - The majority of passengers across all classes boarded here.
#
# 2. Cherbourg (France) - Represented as 'C':
#    - The first stop after leaving England.
#    - A significant number of wealthy First-Class passengers joined the ship here.
#
# 3. Queenstown, modern-day Cobh (Ireland) - Represented as 'Q':
#    - The final stop before the Titanic headed out into the Atlantic Ocean.
#    - Most passengers boarding here were Third-Class emigrants seeking a new life in America.
# ========================================================================================================================================================================
```

## Key Findings & Conclusion

Based on the exploratory data analysis (EDA) of the Titanic dataset, several key insights were identified:

* **Gender Disparity:** There is a significant difference in survival rates between genders. Female passengers had a substantially higher survival rate compared to male passengers, highlighting the "women and children first" policy during the evacuation.
* **Class & Socioeconomic Status:** (If you analyzed this) Passengers in higher classes (1st Class) showed a higher probability of survival compared to those in lower classes.
* **Data Quality:** The cleaning process, including handling missing values in the 'Age' column and optimizing data types, was crucial for ensuring the accuracy of the statistical visualizations.

**Summary:** The survival on the Titanic was not random; it was heavily influenced by demographic factors such as gender, age, and passenger class.
